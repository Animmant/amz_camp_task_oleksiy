```python
# map_sum = { # Pre-calculated values for quick reference
#     1: 1,
#     2: 3,
#     3: 6,
#     4: 10,
#     5: 15,
#     6: 21,
#     7: 28,
#     8: 36,
#     9: 45,
#     10: 55,
#     11: 66,
#     12: 78,
#     13: 91,
#     14: 105,
#     15: 120,
#     19: 190
# }

# res 14, 9, 19, 11

def get_max_floors(k: int, m: int) -> int:
    """
    Calculates the maximum number of floors for k trials and m glasses
    using the sum of the binomial coefficients.
    """
    total = 0
    res = 1
    for i in range(1, m + 1):
        res *= (k - i + 1)
        res //= i
        total += res
        if total >= 10**9: # Fuse for very high values
            break
    return total

def min_attempts_universal(m: int, n: int) -> int:
    k = 0
    while get_max_floors(k, m) < n:
        k += 1
    return k

# Your test cases
test_cases = [(2, 100), (3, 100), (3, 1000), (5, 1000)]

for m, n in test_cases:
    k = min_attempts_universal(m, n)
    print(f"Стаканів: {m}, Поверхів: {n} => Мін. спроб: {k}")
```
